```python
# cook your dish here
import pandas as pd
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, Flatten, Dense, Dropout

# Apply dark mode styling to the generated matplotlib plots
plt.style.use('dark_background')

# Load local CSV files
train = pd.read_csv('mnist_train.csv')
test = pd.read_csv('mnist_test.csv')

# Separate labels and pixels
ytr, xtr = train.iloc[:, 0].values, train.iloc[:, 1:].values
yte, xte = test.iloc[:, 0].values, test.iloc[:, 1:].values

# Reshape and normalize
xtr = xtr.reshape(-1, 28, 28, 1) / 255.0
xte = xte.reshape(-1, 28, 28, 1) / 255.0

print('Train:', xtr.shape)
print('Test:', xte.shape)

# Build model
model = Sequential([
    Input(shape=(28, 28, 1)),
    Conv2D(32, 3, activation='relu'),
    MaxPooling2D(2),
    Conv2D(64, 3, activation='relu'),
    MaxPooling2D(2),
    Flatten(),
    Dense(64, activation='relu'),
    Dropout(0.2),
    Dense(10, activation='softmax')
])

# Compile
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

model.summary()

# Train
history = model.fit(
    xtr, ytr,
    epochs=5,
    batch_size=64,
    validation_split=0.1
)

# Evaluate
loss, acc = model.evaluate(xte, yte)
print('Test Accuracy:', round(acc * 100, 2))

# Predict and display
i = 0
pred = model.predict(xte[i:i+1], verbose=0).argmax()

plt.imshow(xte[i].reshape(28, 28), cmap='gray')
plt.title(f'Predicted: {pred} | Actual: {yte[i]}')
plt.axis('off')
plt.show()
```
