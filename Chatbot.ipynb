```python
import numpy as np
from tensorflow.keras import Sequential
from tensorflow.keras.layers import TextVectorization, Embedding, Bidirectional, LSTM, Dense
from sklearn.preprocessing import LabelEncoder
```

```python
qa = {
    "Hello": "Hi there!",
    "How are you?": "I'm doing great, thank you!",
    "What is your name?": "I am a chatbot using BiLSTM",
    "Bye": "Goodbye, have a nice day!",
    "What do you do?": "I answer your questions based on predefined data"
}
```

```python
questions = np.array(list(qa.keys()))
answers = list(qa.values())

vectorize = TextVectorization(output_mode="int",output_sequence_length=10)
vectorize.adapt(questions)
xtrain = vectorize(questions)

encoder = LabelEncoder()
ytrain = encoder.fit_transform(answers)
```

```python
model = Sequential([
    vectorize,
    Embedding(len(vectorize.get_vocabulary()), 64),
    Bidirectional(LSTM(64)),
    Dense(len(encoder.classes_), activation="softmax")
])

model.compile(optimizer="adam", loss="sparse_categorical_crossentropy")
model.fit(np.array(questions,dtype=object), ytrain, epochs=300, verbose=0)
```

```python
while True:
    text = input("You (quit to stop): ")
    if text.lower() == "quit":
        break
    answer = model.predict(np.array([text],dtype=object), verbose=0).argmax()
    print("Bot:", encoder.inverse_transform([answer])[0])
```
